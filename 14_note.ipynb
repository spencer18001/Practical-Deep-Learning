## PyTorch data loading

參考:
- [Dataloader Design for PyTorch - Tongzhou Wang, MIT](https://www.youtube.com/watch?v=sCsPzVumtR8)

#### [Dataset](https://docs.pytorch.org/docs/stable/data.html#torch.utils.data.Dataset)
- collection of data sample
- map-style dataset: index -> sample
- implements the __getitem__() and __len__() protocols

#### [Sampler](https://docs.pytorch.org/docs/stable/data.html#torch.utils.data.Sampler)
- specify data fetching order
- iterable of dataset indices
- provide an __iter__() method

#### [DataLoader](https://docs.pytorch.org/docs/stable/data.html#torch.utils.data.DataLoader)
- handle data loading logic
- combines a dataset and a sampler, and provides an iterable over the given dataset
- collate into batches: collate_fn
- multi-process loading
  - prefetching mini-batches in parallel
  - `prefetch_factor`: Number of batches loaded in advance by each worker.
  - [Building a Multi-Process Data Loader from Scratch](https://teddykoker.com/2020/12/dataloader/)
