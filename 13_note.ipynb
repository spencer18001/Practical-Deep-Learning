## cross-entropy loss


參考:
- [University of Freiburg - Cross Entropy Loss, KL Divergence and Maximum Likelihood](https://www.youtube.com/watch?v=e-7Tm3YWgT8)

**information theory**

wiki:
```
應用數學、電子學和電腦科學的一個分支，
涉及資訊的量化、儲存和通訊等。
```

**entropy** （熵 ㄕㄤ）
- measure of uncertainty, lack of predictability
- \# bits to code an event
- $H(P) = -\mathbb{E}_{X \sim P}[\log P(X)]$

gpt:
```
一個只會產生 "A" 的資料來源：
  熵 = 0（因為你完全知道下一個字是什麼，沒有資訊）

一個有 50% 機率產生 "A"，50% 機率產生 "B" 的資料來源：
  熵 = 1 bit（因為你不確定它會傳什麼，每次都提供 1 bit 的資訊）
```
gpt:
```
熵 和 變異數 雖然都是衡量「不確定性」或「分散程度」的指標，
但它們的關係並不是簡單的正比或等價關係，取決於使用的機率分布類型。

熵考慮的是「整體分布的不確定性」，而變異數僅衡量「平均值附近的分散程度」。
```

**cross-entropy**
- $H(P,Q) = -\mathbb{E}_{X \sim P}[\log Q(X)]$
- using a code that is optimized for the wrong distribution $Q(x)$
- minimized when $P=Q$

gpt:
```
若你高估某事件（以為很常發生），實際卻很少發生，
→ 你會給它短的編碼，但它實際上出現時會造成困擾（不夠代表性）。

若你低估某事件（以為很少發生），實際卻常發生，
→ 你會給它長的編碼，而它卻常出現 → 平均訊息長度被拉高。
```

**KL divergence**
- $\begin{align}
D_{\mathrm{KL}}(P \,\|\, Q)
&= -\mathbb{E}_{X \sim P}\left[\log \left( \frac{Q(X)}{P(X)} \right) \right] \\
&= H(P, Q) - H(P)
\end{align}$
- distance metric of distributions
- not symmetric
- extra number of bits

**MLE minimizes cross-entropy, KL divergence:**

$\begin{align*}
\theta_{\text{ML}}
&= \arg\max_{\theta} p_{\text{model}}(\mathbb{X}; \theta) \\
&= \arg\max_{\theta} \prod_{i=1}^{m} p_{\text{model}}(x^{(i)}; \theta) \\
&= \arg\max_{\theta} \sum_{i=1}^{m} \log p_{\text{model}}(x^{(i)}; \theta) \\
&= \arg\max_{\theta} \mathbb{E}_{x \sim \hat{p}_{\text{data}}} [\log p_{\text{model}}(x; \theta)] \\
&= \arg\min_{\theta} \mathbb{E}_{x \sim \hat{p}_{\text{data}}} [- \log p_{\text{model}}(x; \theta)] \\
&= \arg\min_{\theta} H(\hat{p}_{\text{data}}, p_{\text{model}}(\cdot; \theta)) \quad \text{(cross entropy)} \\
&= \arg\min_{\theta} \mathbb{E}_{x \sim \hat{p}_{\text{data}}} [\log \hat{p}_{\text{data}}(x) - \log p_{\text{model}}(x; \theta)] \\
&= \arg\min_{\theta} D_{\mathrm{KL}}(\hat{p}_{\text{data}}(\cdot) \| p_{\text{model}}(\cdot; \theta)) \quad \text{(KL divergence)}
\end{align*}$

**cross-entropy loss**
- $\mathcal{L}(\mathbf{y}, \hat{\mathbf{y}}) = -\sum_{k=1}^{K} y_k \log \hat{y}_k$

gpt:
```
如果模型預測越接近正確類別的機率為 1，cross entropy 趨近於 0。
如果預測錯得離譜（把機率給了錯誤類別），cross entropy 就很大。
```
```
每筆樣本的 cross entropy loss 是整體 cross entropy 的 sample-level 表達
```
