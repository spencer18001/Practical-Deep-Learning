## Classifier-Free Guidance

vanilla conditional generation

image.png

*Tweedie’s formula*

Optimal denoising strategy is to follow the gradient (score) (of the log-perturbed data density) $\nabla_\tilde{x} \log q_\sigma(\tilde{x} \mid x)$

- $\nabla_x \log p(x \mid y)=\nabla_x \log p(x) + \lambda \left( \nabla_x \log p(x \mid y) - \nabla_x \log p(x) \right)$
- `noise_pred = noise_pred_uncond + guidance_scale * (noise_pred_text - noise_pred_uncond)`
